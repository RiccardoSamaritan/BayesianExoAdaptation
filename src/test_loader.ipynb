# Test Loader

This notebook tests the loading of HAR data using the `loader.py` module.

## 1. Module Import

```python
import sys
import numpy as np
from loader import load_har, validate, HARDataset, ACTIVITY_ID_TO_NAME, LOCOMOTION_IDS

print("✓ Import successful!")
```

## 2. Dataset Loading

```python
# Load complete dataset (train + test merged)
dataset = load_har("../data", merge_train_test=True)

print(f"✓ Dataset loaded!")
print(f"  - X shape: {dataset.X.shape}")
print(f"  - y shape: {dataset.y.shape}")
print(f"  - subject_id shape: {dataset.subject_id.shape}")
print(f"  - Number of features: {len(dataset.feature_names)}")
```

## 3. First Rows Visualization

```python
print("First 5 rows of features:")
print(dataset.X[:5, :5])  # First 5 rows, first 5 columns
print("\nFirst 10 activity labels:")
print(dataset.y[:10])
print("\nFirst 10 subject IDs:")
print(dataset.subject_id[:10])
```

## 4. Activity Name Property Test

```python
print("First 10 activities by name:")
print(dataset.activity_name[:10])
print("\nAvailable activity mapping:")
for id, name in ACTIVITY_ID_TO_NAME.items():
    print(f"  {id}: {name}")
```

## 5. Dataset Validation

```python
validation_report = validate(dataset)

print("Validation report:")
print("=" * 50)
for key, value in validation_report.items():
    print(f"{key:25s}: {value}")
```

## 6. Activity Filtering Test

```python
# Filter only locomotion activities (WALKING, WALKING_UPSTAIRS, WALKING_DOWNSTAIRS)
print(f"Locomotion activity IDs: {LOCOMOTION_IDS}")
locomotion_dataset = dataset.filter_activities(LOCOMOTION_IDS)

print(f"\nComplete dataset: {dataset.X.shape[0]} samples")
print(f"Filtered dataset (locomotion only): {locomotion_dataset.X.shape[0]} samples")
print(f"\nActivities present in filtered dataset:")
print(np.unique(locomotion_dataset.activity_name))
```

## 7. Subject Filtering Test

```python
# Filter only first 5 subjects
subset_subjects = [1, 2, 3, 4, 5]
subject_dataset = dataset.filter_subjects(subset_subjects)

print(f"Complete dataset: {dataset.X.shape[0]} samples")
print(f"Filtered dataset (subjects {subset_subjects}): {subject_dataset.X.shape[0]} samples")
print(f"\nSubjects present in filtered dataset:")
print(np.unique(subject_dataset.subject_id))
```

## 8. Combined Test: Activity and Subject Filtering

```python
# Combine filters: only locomotion for first 5 subjects
combined_dataset = dataset.filter_activities(LOCOMOTION_IDS).filter_subjects([1, 2, 3, 4, 5])

print(f"Complete dataset: {dataset.X.shape[0]} samples")
print(f"Dataset with combined filters: {combined_dataset.X.shape[0]} samples")
print(f"\nActivities:")
print(np.unique(combined_dataset.activity_name))
print(f"\nSubjects:")
print(np.unique(combined_dataset.subject_id))
```

## 9. Feature Names Visualization

```python
print(f"Total number of features: {len(dataset.feature_names)}")
print("\nFirst 20 features:")
for i, name in enumerate(dataset.feature_names[:20]):
    print(f"  {i+1:3d}. {name}")
```

## 10. Class Distribution

```python
print("Class distribution:")
print("=" * 50)
unique_activities, counts = np.unique(dataset.y, return_counts=True)
for activity_id, count in zip(unique_activities, counts):
    activity_name = ACTIVITY_ID_TO_NAME[activity_id]
    percentage = (count / len(dataset.y)) * 100
    print(f"{activity_name:25s}: {count:5d} ({percentage:5.2f}%)")
print("=" * 50)
print(f"{'TOTAL':25s}: {len(dataset.y):5d} (100.00%)")
```

## 11. Subject Distribution

```python
unique_subjects = np.unique(dataset.subject_id)
print(f"Number of subjects: {len(unique_subjects)}")
print(f"Subject ID range: {unique_subjects.min()} - {unique_subjects.max()}")
print("\nSamples per subject:")
for subject_id in sorted(unique_subjects):
    count = (dataset.subject_id == subject_id).sum()
    print(f"  Subject {subject_id:2d}: {count:4d} samples")
```
